**Koneksi ke MongoDB dari Python**

```python
from pymongo import MongoClient

# Membuat koneksi ke MongoDB yang berjalan di localhost, port default 27017
client = MongoClient("mongodb://localhost:27017/")

# Memilih (atau otomatis membuat) database
db = client["kopi_nusantara"]

# Memilih (atau otomatis membuat) collection
data_menu = db["menu"]

print("Koneksi ke MongoDB berhasil!")
print("Daftar database yang ada:", client.list_database_names())
```

### **A. Membangun Katalog (Create)**

```python
data_menu.delete_many({})

data_menu.insert_many([
    {"nama_menu":"Espresso", "harga":18000, "kategori":"Kopi", "tingkat_kepahitan":5, "stok":20, "rekomendasi_barista":True},
    {"nama_menu":"Americano", "harga":20000, "kategori":"Kopi", "tingkat_kepahitan":4, "stok":25, "rekomendasi_barista":True},
    {"nama_menu":"Cappuccino", "harga":27000, "kategori":"Kopi", "tingkat_kepahitan":3, "stok":18, "rekomendasi_barista":True},
    {"nama_menu":"Cafe Latte", "harga":28000, "kategori":"Kopi", "tingkat_kepahitan":2, "stok":23},
    {"nama_menu":"Kopi Tubruk", "harga":15000, "kategori":"Kopi", "tingkat_kepahitan":5, "stok":27},
    {"nama_menu":"Aren Latte", "harga":25000, "kategori":"Kopi", "tingkat_kepahitan":2, "stok":30, "rekomendasi_barista":True},
    {"nama_menu":"Cold Brew", "harga":30000, "kategori":"Kopi", "tingkat_kepahitan":4, "stok":15},
    {"nama_menu":"Mocha", "harga":29000, "kategori":"Kopi", "tingkat_kepahitan":3, "stok":16},
    {"nama_menu":"Nasi Goreng Nusantara", "harga":25000, "kategori":"Makanan", "pedas_level":2, "stok":30},
    {"nama_menu":"Mie Goreng Jawa", "harga":20000, "kategori":"Makanan", "pedas_level":2, "stok":21},
    {"nama_menu":"Pisang Goreng", "harga":13000, "kategori":"Makanan", "pedas_level":0, "stok":15},
    {"nama_menu":"Roti Panggang Coklat", "harga":13000, "kategori":"Makanan", "pedas_level":0, "stok":16},
])

print("Total dokumen: ", data_menu.count_documents({}))
```

### **B. Kueri Bertingkat (Read)**

**Seluruh menu kategori "Kopi" dengan tingkat_kepahitan di atas 3.**

```python
for data in data_menu.find({"kategori":"Kopi", "tingkat_kepahitan":{"$gt":3}},{"_id":0}):
    print(data)
```

**Seluruh menu dengan harga di antara 15000 dan 30000.**

```python
for data in data_menu.find({"harga":{"$gte":15000, "$lte":30000}},{"_id":0}):
    print(data)
```

**Menu yang tidak memiliki field pedas_level sama sekali.**

```python
for data in data_menu.find({"pedas_level":{"$exists":False}},{"_id":0}):
    print(data)
```

### **C. Pembaruan Data (Update)**

```python
hasil = data_menu.update_many({},{"$set":{"status":"Tersedia"}})

data_menu.update_one(
    {"nama_menu":"Espresso"},
    {"$set":{"status":"Habis"}}
)
data_menu.update_one(
    {"nama_menu":"Nasi Goreng Nusantara"},
    {"$set":{"status":"Habis"}}
)

for data in data_menu.find({}, {
    "_id":0,
    "nama_menu":1,
    "status":1,
}):
    print(data)
```

### **D. Integrasi ke Spark**

```python
from pyspark.sql import SparkSession
import pandas as pd

spark = SparkSession.builder \
    .appName("Tugas9-KopiNusantara") \
    .master("local[*]") \
    .getOrCreate()
spark.sparkContext.setLogLevel("ERROR")

dokumen = list(data_menu.find({}))

df_pandas = pd.DataFrame(dokumen)
df_pandas["_id"] = df_pandas["_id"].astype(str)
df_pandas = df_pandas.astype(object).where(
    pd.notna(df_pandas), None
)

df_spark = spark.createDataFrame(df_pandas)
df_spark.printSchema()
df_spark.groupBy("kategori").avg("harga").show()
```

### **E. Kesimpulan**

Tantangan utama ketika mengonversi data MongoDB ke Spark DataFrame adalah struktur dokumen MongoDB yang fleksibel. Setiap dokumen dapat memiliki field yang berbeda. Contohnya, menu kategori Kopi mempunyai field *tingkat_kepahitan*, sedangkan menu kategori Makanan mempunyai field *pedas_level*. Akibatnya, beberapa dokumen tidak mempunyai nilai untuk field tertentu. 

Untuk mengatasi hal tersebut, data diambil dari MongoDB menggunakan PyMongo, kemudian dikonversi menjadi Pandas DataFrame sebelum diubah menjadi Spark DataFrame. Field *_id* diubah menjadi string agar tipe ObjectId kompatibel dengan proses inferensi skema Spark. Nilai kosong juga ditangani sebagai None, sehingga Spark dapat merepresentasikannya sebagai null. Setelah dikonversi, struktur data diperiksa menggunakan *printSchema()*. Selanjutnya, operasi *groupBy()* digunakan untuk mengelompokkan menu berdasarkan kategori dan menghitung rata-rata harganya. Dengan demikian, fleksibilitas penyimpanan MongoDB tetap dapat dimanfaatkan. Sedangkan Spark digunakan untuk melakukan analisis data secara terstruktur.
